# Removing elements

`view.remove(...)` removes atoms and/or structures from the molecular system behind the view.

This is a live operation: it updates `view.molsys` and refreshes the viewer payload.

A practical guideline: prefer removing by **selection** rather than by hard-coded atom indices, because atom indices can change after removals.


```python
import molsysviewer as viewer

view = viewer.demo["181L"]
view.show()
```

```python
from molsysviewer.thirds.jupyter import load_html_in_notebook

load_html_in_notebook("../../../_static/views/demo_181L.html")
```

## Example: remove water

We start by checking how many atoms we have, and then we remove water molecules.


```python
view.get(element="system", n_atoms=True, n_waters=True, output_type="dictionary")
```

```python
view.remove(selection='molecule_type=="water"')
view.get(element="system", n_atoms=True, n_waters=True, output_type="dictionary")
```
