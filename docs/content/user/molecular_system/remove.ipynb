# Removing elements and structures

Calculate removals with MolSysMT, then load the resulting system.

MolSysMT is the native scientific backend. Compute a modified system with `molsysmt.remove(...)` and replace the scene with `view.load(molsys, mode="replace")`. Replacement resets scene objects. Advanced integrations can retain them through `view.apply_system_edit(...)` with explicit index maps.

For more details and additional examples, see the MolSysMT tutorial: [Remove](https://www.uibcdf.org/molsysmt/content/user/tools/basic/remove).

A practical guideline: whenever you can, prefer removing by **selection** rather than by hard-coded atom indices. Atom indices can change after removals, and selections are usually more robust.


## Elements and structures

`remove` supports two complementary operations, and you can use either one depending on what you are trying to do:

- Remove atoms (and therefore any higher-level elements that become empty) using `selection=...`.
- Remove structures using `structure_indices=...`.

## Example: remove water

We start by checking how many atoms we have, and then we remove water molecules.

This is a destructive operation on the loaded system: if you want to undo it, reload the view (or re-create it from the original data).

````python
import molsysviewer as msv
import molsysmt as msm


view = msv.demo["181L"]
view.show()
````

````python
from molsysviewer.thirds.jupyter import load_html_in_notebook

load_html_in_notebook("../../../_static/views/demo_181L.html")
````

````python
msm.get(view, element="system", n_atoms=True, n_waters=True, output_type="dictionary")
````

````python
molsys = msm.remove(view.molsys, selection='molecule_type=="water"', to_form="molsysmt.MolSys")
view.load(molsys, mode="replace")
msm.get(view, element="system", n_atoms=True, n_waters=True, output_type="dictionary")
````

## Example: remove structures

Some systems include multiple structures (for example, a trajectory or multiple coordinate sets). In that case, you can remove structures by index (0-based).

In the demo below, the molecular system includes many structures, and we remove a few of them.

````python
view = msv.demo["pentalanine"]
view.show()
````

````python
load_html_in_notebook("../../../_static/views/demo_pentalanine.html")
````

````python
msm.get(view, element="system", n_structures=True)
````

````python
molsys = msm.remove(view.molsys, structure_indices=[0, 1, 2], to_form="molsysmt.MolSys")
view.load(molsys, mode="replace")
msm.get(view, element="system", n_structures=True)
````
