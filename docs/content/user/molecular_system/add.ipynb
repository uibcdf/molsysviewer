# Adding elements

`view.add(...)` adds atoms and structures from another molecular system into the current view.

This is a live operation: it mutates the system behind the view and refreshes the viewer.

You typically use it when you want to combine two systems into one (for example, adding a small molecule system into a protein system).


```python
import molsysviewer as viewer

view = viewer.demo["181L"]
view.show()
```

```python
from molsysviewer.thirds.jupyter import load_html_in_notebook

load_html_in_notebook("../../../_static/views/demo_181L.html")
```

## Example: add another system

Here we add a small demo system into the current one, and we check that the number of atoms increases.


```python
before = view.get(element="system", n_atoms=True)
before
```

```python
other = viewer.demo["dialanine"].molsys
view.add(other)
after = view.get(element="system", n_atoms=True)
before, after
```
