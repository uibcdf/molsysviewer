# Appending structures

`view.append_structures(...)` appends additional structures to the current molecular system.

This is most commonly used to extend a molecular dynamics trajectory (for example, when you want to combine two trajectory chunks of the same system).

This is a live operation: it mutates the system behind the view and refreshes the viewer.


```python
import molsysviewer as viewer

view = viewer.demo["pentalanine"]
view.show()
```

```python
from molsysviewer.thirds.jupyter import load_html_in_notebook

load_html_in_notebook("../../../_static/views/demo_pentalanine.html")
```

## Example: append a few structures

We start from a trajectory demo system, check how many structures it has, append a few more structures, and check the new count.

In real workflows, the system you append from should represent the same topology.


```python
before = view.get(element="system", n_structures=True)
before
```

```python
other = viewer.demo["pentalanine"].molsys
view.append_structures(other, structure_indices=[0, 1, 2])
after = view.get(element="system", n_structures=True)
before, after
```
