# Appending structures

This page documents `view.load(..., mode="append_structures")`.

Use this loading mode to append additional structures to the current molecular system. If your molecular system represents a trajectory, structures are often called "frames". Appending structures is the most common way to extend a trajectory with an additional chunk.

`mode="append_structures"` is loader sugar: it loads the additional structures and reconciles the viewer state. For MolSysMT-owned live editing, use `view.addons.molsysmt.basic.append_structures(...)` from the MolSysMT addon.

For more details and additional examples, see the MolSysMT tutorial: [Append structures](https://www.uibcdf.org/molsysmt/content/user/tools/basic/append_structures).


## Example: append a second structure

In {doc}`add`, we combined two systems by increasing the number of atoms.

Here we do something different: we keep the same atoms and topology, but we add a second set of coordinates. The result is a molecular system with two structures.

A key requirement is that the system you append from represents the same topology. In this example, we guarantee that by appending from a translated copy of the same system.

```python
import molsysviewer as viewer
import molsysmt as msm

view = viewer.demo["dialanine"]
msm.get(view, element="system", n_structures=True)
```

```python
view.show()
```

```python
from molsysviewer.thirds.jupyter import load_html_in_notebook

load_html_in_notebook("../../../_static/views/demo_dialanine.html")
```

Now let's add the second structure:

```python
import molsysmt as msm
import pyunitwizard as puw

second_structure = msm.copy(view.molsys)
coordinates = msm.get(second_structure, element="atom", coordinates=True)
msm.set(
    second_structure,
    coordinates=coordinates + puw.quantity([0.2, 0.0, 0.0], "nanometer"),
)

view.load(second_structure, mode="append_structures")
msm.get(view, element="system", n_structures=True)
```
