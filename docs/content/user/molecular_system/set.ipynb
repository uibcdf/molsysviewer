# Setting attributes

Calculate attribute changes with MolSysMT, then load the modified system.

MolSysMT is the native scientific backend. Edit a copy with `molsysmt.set(...)` and replace the scene with `view.load(molsys, mode="replace")`. Replacement resets scene objects. Advanced integrations can retain them through `view.apply_system_edit(...)` with explicit index maps.

`set` follows MolSysMT's attribute names and selection rules.

For more details and additional examples, see the MolSysMT tutorial: [Set](https://www.uibcdf.org/molsysmt/content/user/tools/basic/set).

## Example: rename a group

Here we rename a single group (by `group_index`) and then revert the change.

This is a simple example, but it shows the mental model: select what you want to modify, then set a new value.

We start by retrieving the current group name so we can safely restore it at the end.

````python
old_name = msm.get(view, element="group", selection="group_index==30", group_name=True)[0]
old_name
````

````python
molsys = msm.copy(view.molsys)
msm.set(molsys, selection="group_index==30", group_name="HSD")
view.load(molsys, mode="replace")
msm.get(view, element="group", selection="group_index==30", group_name=True)
````

## A second example: split chain names

In MolSysMT/MolSysViewer, a “chain” is a topological level that may depend on how the input was authored.

In this demo, all chains start with the same chain name (`"A"`). We will assign the name `"B"` to every chain whose index is not 0.

This is a convenient way to quickly separate the main polymer chain from everything else in the metadata.

````python
msm.get(view, element="chain", chain_index=True, chain_id=True, chain_name=True, output_type="dictionary")
````

````python
molsys = msm.copy(view.molsys)
msm.set(molsys, element="chain", selection="chain_index!=0", chain_name="B")
view.load(molsys, mode="replace")
````

````python
msm.get(view, element="chain", chain_index=True, chain_id=True, chain_name=True, output_type="dictionary")
````
