# Installation

This page walks you through installing MolSysViewer and checking that it works.

## Install with conda

MolSysViewer is distributed through the `uibcdf` conda channel.
For most users, conda is the easiest and most reliable way to get a working setup.

If you are starting fresh, we recommend creating a dedicated conda environment so
your Jupyter stack and scientific dependencies stay consistent. For example:

```bash
conda create -n molsysviewer-env -c uibcdf -c conda-forge -c ambermd python=3.14 molsysviewer
conda activate molsysviewer-env
```

If you already have an environment you want to use, just install the package:

```bash
conda install -c uibcdf -c conda-forge -c ambermd molsysviewer
```

You do not need to manually manage individual dependencies. Conda will resolve and
install the required packages for you.

Under the hood, MolSysViewer relies on `anywidget` (for the Jupyter widget layer), and
it is designed to work smoothly with MolSysMT and PyUnitWizard. When you use conda,
you do not need to install these dependencies manually.

If you plan to work in Jupyter, we recommend installing Jupyter in the same environment
and launching it from there. For example:

```bash
conda install -c conda-forge jupyterlab
```

## Platform and host scope

| Capability | Platform and Python scope |
| --- | --- |
| Core Python API, Jupyter widget and interactive HTML | Linux x86_64/ARM64, macOS Apple Silicon, Windows x86_64; Python 3.11–3.14 |
| Local standalone launchers and Qt desktop host | Experimental for 1.0; native Qt dependencies and visible rendering require separate qualification |
| Remote sessions | Unsupported preview; supported workflows remain post-1.0 |

MolSysMT is the required backend, not an addon you must enable. Conda is the
public installation route; MolSysViewer is not distributed through PyPI.
These public installed results are separate from central Python 3.14 admission,
which remains pending under `uibcdf/molsysviewer#93`. Intel-based macOS is outside
the current matrix. Installing the core noarch
package does not certify a native Qt host. The experimental Qt recipe has Linux
transport evidence and Windows/macOS solver-only observations; see
{doc}`../../developer/standalone_surfaces` for its separate host contract.
