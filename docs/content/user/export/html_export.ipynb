# HTML Export

Once you have the view in the state you want (loaded structure, chosen representations, camera orientation, etc.), you can export it to a standalone HTML file using the `export.html` method. This lets you embed an interactive version of the view in your own web pages.

The exported file contains:

- the MolSysViewer runtime — the renderer itself,
- the scene, as the sequence of actions that produced it (loads, visibility
  changes, colours, regions, camera) up to the moment you call `export.html`.

Opening this file in a browser recreates the same interactive view. It needs no
Python kernel, no server and no network: everything is in the file.

To show it, let's first of all create a simple view.

````python
import molsysviewer as viewer
from molsysviewer import new_view
````

## Exporting the current view

The method `export.html` writes an HTML file containing the interactive view:

```python
view.export.html(
    output_filename="embedding_test.html",
    title="Embedding Test",
)
```

- `output_filename` is the path of the HTML file to create.
- `title` is optional and sets the `<title>` of the HTML page.

By default, the exported view includes the on-canvas control buttons (Reset, Fullscreen, background toggle, Spin, Swing) so that readers of your documentation can manipulate the scene.

After the file is written, you can open it directly with your browser (for example, by double-clicking on it in your file manager or serving it from a simple web server).

## Shared or self-contained?

One argument decides the shape of the output.

|  | no `shared_runtime` (default) | `shared_runtime="<dir>"` |
|---|---|---|
| Runtime | inside every file | shared, one copy for all views |
| File size | several MB each | small |
| Opens by double click | yes | no — must be served |
| Works with no network | yes | yes |
| Best for | a single file you send to someone | a site with more than one view |

```python
view.export.html("report.html")                                  # self-contained
view.export.html("docs-embed.html", shared_runtime="docs/_static")  # shared
```

The shared copy is the runtime installed with your MolSysViewer, placed in that
directory — version-exact with the scene, and working offline once the site is
served. A browser will not let a page opened straight from a disk load the file
next to it, so use `python -m molsysviewer.preview docs/_build/html` while
you work. `"cdn"` and an explicit list of URLs are the two further values; see
{doc}`sphinx_html_embedding` for the full workflow.

## Full signature

```python
view.export.html(
    output_filename,           # path to write
    title="MolSysViewer",      # HTML <title> tag
    include_controls=True,     # on-canvas control buttons
    include_popout=True,       # popout-window button
    shared_runtime=None,       # None, a directory, "cdn", or a list of URLs
    inline_messages=True,      # shared only: embed messages inline vs external JSON
)
```

## Exporting without displaying the widget

You do not need to call `show()` in order to export the view. This is useful when you are generating views from a script or a non-interactive environment:

```python
from molsysviewer import new_view

view = new_view("1TCD")
# Optionally perform additional operations here (load more systems, change visibility, etc.)
view.export.html("my_view.html", title="My MolSysViewer scene")
```

The resulting `my_view.html` file is again a standalone HTML page that you can open in any modern browser.

## Hiding the built-in controls

If you prefer a minimal viewer (for example, when embedding it inside a larger application that already provides its own UI), you can disable the MolSysViewer control buttons in the exported HTML:

```python
view.export.html(
    "my_view_no_toolbar.html",
    title="My MolSysViewer scene",
    include_controls=False,
)
```

Setting `include_controls=False` hides the on-canvas toolbar (Reset, Fullscreen, background toggle, Spin, Swing) in the generated HTML, while keeping the scene itself fully interactive.

## Embedding in other documentation systems

An exported view is an ordinary HTML file, so you can reuse it anywhere:

- link to it from your documentation,
- serve it from a static web server,
- embed it in another page with an `<iframe>`.

A file exported without `shared_runtime` carries everything it needs and opens
with a double click. One exported with it is accompanied by a single shared
runtime asset, which is what makes it small enough for a site with many views —
and which has to be reached over HTTP, so serve it.

{doc}`sphinx_html_embedding` covers the whole workflow — where the runtime comes
from, how to share one copy between views, and how to embed the result in Sphinx,
MkDocs, Quarto or plain HTML.
