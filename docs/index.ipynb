## Install it

```bash

```

<br/>

## Use it

````python
import molsysviewer as viewer

view = viewer.load('1TCD')
view.info()
````

````python
view.clean()
view.show(selection='all within 4.5 angstroms of atom_name=="CA" and residue_id=="23"')
````
