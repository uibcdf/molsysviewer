## Use it

````python
import molsysviewer as msv

view = msv.new_view('181L', selection='molecule_type not in ["water", "ion"]')
view.info(element='molecule')
````

Note: `new_view` defaults to `load_mode="selection"`, so the selection subsets the system. Use `load_mode="all"` to load the full system and show the selection as a region.


````python
region_1 = view.regions.add('all within 9 angstroms of molecule_name=="BENZENE"', representation='sticks')
view.zoom('molecule_name=="BENZENE"')
view.show()
````

````python
from molsysviewer.thirds.jupyter import load_html_in_notebook
load_html_in_notebook('_static/views/181L_benzene.html')
````
