# Unit 06: Programmatic Selection

Using MolSysMT queries to select structural subsets programmatically.

### Learning Objectives
1. Use standard MolSysMT selection keywords (backbone, sidechain, etc.).
2. Perform shell selection queries (within a radius).
3. Map selection strings to list of atom indices.

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Selecting Structural Elements

We use `view.select()` to query atom indices. Let's retrieve all backbone atoms.


```python
# Select backbone atoms
backbone_indices = view.select(selection='backbone')
print(f"Backbone has {len(backbone_indices)} atoms. Indices:", backbone_indices[:10])
```

## 3. Selecting Shells around residues

We can query atoms that lie within a specific physical distance from a target.


```python
# Select all atoms within 5.0 Angstroms of residue index 10
pocket_indices = view.select(selection='all within 5.0 angstroms of group_index==10')
print(f"Found {len(pocket_indices)} atoms in pocket shell.")
```
