# Unit 15: Precision Measurements

Measuring distances, angles, and dihedrals using physical units.

### Learning Objectives
1. Create distance, angle, and dihedral measurements.
2. Integrate physical quantities with PyUnitWizard.
3. Extract values as unit-safe quantities.

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Adding Measurements

We can measure geometric parameters by passing atom index queries.


```python
# Measure distance between atom 10 and 50
view.measurements.add_distance(atom_indices=[10, 50], tag='dist_1')

# Measure angle between atoms 10, 11, and 12
view.measurements.add_angle(atom_indices=[10, 11, 12], tag='angle_1')
```

## 3. Retrieving Values as Quantities

We can fetch the measured values in physical units.


```python
# Get detailed records
records = view.measurements.info()
print("Measurement results:", records)
```
