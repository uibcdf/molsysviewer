# Unit 15: Professional Export

A great visualization is only useful if you can share it. In this final unit of the production block, we will learn how to immortalize our findings. We will explore high-resolution image capture, the use of **Figure Recipes** for consistent publication styles, and the generation of standalone HTML sessions.

### Learning Objectives
1. Generate high-resolution PNG images with transparent backgrounds.
2. Master **FigureSpec** to create consistent figure "recipes".
3. Export interactive, standalone HTML files for sharing.

## 1. Simple Image Export

The `view.export.image()` method captures the current state of the canvas. You can scale the image (e.g., 2.0 for double resolution) to meet journal requirements.

```python
import molsysviewer as msv
view = msv.MolSysView()
view.load('1CRN')

# Note: The viewer must be rendered at least once before exporting
view

# Export a high-quality PNG
view.export.image(
    'my_protein.png', 
    scale=2.0, 
    transparent=True
)
```

## 2. Consistent Figures with `FigureSpec`

For professional publications, you often need multiple figures to share the exact same style (background, lighting, quality). A `FigureSpec` acts as a reusable "recipe".

```python
from molsysviewer.figures import FigureSpec

# Define a standard recipe for your lab's publications
journal_style = FigureSpec(
    background='white', 
    preset='publication-light', 
    scale=3.0
)

# Apply the recipe to the current view
view.export.image('figure_1.png', figure_spec=journal_style)
```

## 3. Standalone HTML Sessions

Sometimes, a static image is not enough. You can export the entire interactive session as an HTML file. This file contains the molecular data and the viewer code, meaning anyone can open it in a browser without needing Python or MolSysViewer installed.

```python
# Export the interactive scene
view.export.html(
    'interactive_session.html', 
    title="Crambin Analysis", 
    mode='standalone'
)
```

### Summary
Professional export ensures that your discoveries are communicated clearly and consistently.
- Use `image()` with high `scale` for publications.
- Use `FigureSpec` to automate your lab's visual standards.
- Use `html(mode='standalone')` for truly portable scientific sharing.
