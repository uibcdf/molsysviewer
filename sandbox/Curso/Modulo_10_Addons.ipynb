Módulo 10: Superpoderes Moleculares (Addons y Extensibilidad)

Para este módulo (Módulo_10_Addons.ipynb), el alumno aprenderá que MolSysViewer puede crecer instalando módulos de la comunidad o propios.

  Objetivos Pedagógicos:
   1. Listar los addons disponibles.
   2. Activar un addon específico.
   3. Ver cómo la interfaz del visor cambia al activar un addon.


```python
import molsysviewer as msv
```

```python
print(f"Addons disponibles: {msv.addons.available()}")
```

```python
from molsysviewer.addon_templates import minimal_elasnetmt
```

```python
msv.addons.register_module(minimal_elasnetmt)
```

```python
print(f"Disponibles ahora: {msv.addons.available()}")
```

```python
view = msv.MolSysView()
```

```python
view.addons.enable('elasnetmt-template')
```

```python
print(f"Addon activo: {view.addons.is_enabled('elasnetmt-template')}")
```
