# Unit 03: Simulation Box & Projections

Visualizing periodic boundary conditions and choosing camera projection modes.

### Learning Objectives
1. Render unit-cell or simulation-box edges in the canvas.
2. Customize the width, color, and transparency of the box boundaries.
3. Toggle between technical (Orthographic) and immersive (Perspective) camera projection modes.

## 1. Setting the Stage: Loading Crambin

First, we load Crambin (PDB ID: `1CRN`), which contains unit cell crystallographic box parameters.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Interactive UX Action: Toggle the Simulation Box

Try clicking on the background context menu (or checking the Workbench panel) to display the simulation unit cell. Observe how a grey outline representing the unit cell dimensions appears around the Crambin structure.


## 3. Programmatic API Equivalent: Drawing and Customizing the Box

We can draw, style, and hide the box boundaries using the `show_box()` and `hide_box()` methods.


```python
# Render a custom semi-transparent red box
view.show_box(color='red', width=0.1, alpha=0.6)
```

```python
# Hide the box
view.hide_box()
```

## 4. Camera Projection Modes: Perspective vs. Orthographic

By default, the camera displays the scene in **Perspective** mode, which gives a realistic sense of depth. For flat technical layouts, we can programmatically switch to **Orthographic** mode.


```python
# Switch to flat orthographic view
view.camera.set_mode('orthographic')
```

```python
# Switch back to perspective
view.camera.set_mode('perspective')
```
