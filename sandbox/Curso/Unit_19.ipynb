# Unit 19: The Molecular Black Box (SMonitor)

Modern molecular visualization is a complex orchestra of data processing and GPU rendering. When something goes wrong—or when you simply want to understand the sequence of events—you need a "Black Box". In this unit, we will learn how to use the **SMonitor** integration to audit and troubleshoot our session.

### Learning Objectives
1. Inspect the live signal timeline of the viewer.
2. Use the `report()` method to audit events and performance.
3. Understand how to generate diagnostic data for bug reporting.

## 1. Capturing the Signal Trail

Every action you take in **MolSysViewer** (loading, styling, clicking) emits a signal. These signals are tracked by a specialized layer called `smonitor`.

```python
import molsysviewer as msv
view = msv.MolSysView()
view.load('1CRN')

# Perform some actions to generate signals
view.camera.zoom(selection='res_id 10')
view.whole.set_representation(representation='licorice')
view
```

## 2. Auditing with `report()`

The `report()` method provides a structured summary of everything that has happened in your session. It is the primary tool for "forensic" structural biology.

```python
# Generate a session report
audit = view.report()

# Look at the timeline of events
print("Recent Signals:")
for event in audit['timeline'][-5:]: # Show last 5 signals
    print(f"- {event['signal']} (Source: {event['source']})")
```

## 3. Performance and Timings

SMonitor also tracks how long each operation takes. This is useful for identifying bottlenecks when working with massive systems or trajectories.

```python
# Inspect performance metadata
timings = audit.get('timings_by_tag', {})
print(f"Performance Snapshot: {timings}")
```

### Summary
SMonitor ensures that your visualization session is traceable and reproducible.
- Use `view.report()` to see the event history.
- Audit the `timeline` to troubleshoot failed operations.
- Use these tools to provide clear evidence when reporting issues to developers.
