# Unit 19: Clipping and Sectioning

Slicing through structures using clipping planes.

### Learning Objectives
1. Configure camera Z-clipping limits.
2. Add world-space section planes.
3. Toggle interactive Gizmo drag handles.

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Camera Z-Clipping

Z-clipping slices the front or back of the camera viewpoint.


```python
# Clip the front 40% of the scene
view.scene.set_clip_planes(near=40.0)
```

```python
# Reset camera clipping
view.scene.set_clip_planes(near=0.0)
```

## 3. Adding Cutting Planes (Sections)

We can add a physical cutting plane through any centroid coordinates.


```python
# Add cutting plane with Z-axis normal
my_cut = view.scene.add_section(point=[0,0,0], normal=[0,0,1], tag='core_slice')

# Disable gizmo drag handles
my_cut.disable_drag()
```
