# Unit 04: Styles and Representations

In this unit, we will explore the "visual atoms" of **MolSysViewer**: the individual representations. You will learn how to manually control how every atom and bond is drawn, and how to apply these styles to specific parts of your molecule using **Regions**.

### Learning Objectives
1. Control the global appearance using the `whole` object.
2. Master the core representation types (`cartoon`, `licorice`, `spacefill`).
3. Fine-tune styles using parameters like **alpha** (transparency).
4. Create and style **Regions** for surgical visual control.

## 1. Setting the Stage

Let's start by loading our reference protein. By default, MolSysViewer applies a standard 'cartoon' representation.

```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Global Styles with `whole`

The `view.whole` object defines the baseline look for the entire system. We can switch styles or adjust transparency globally.

```python
# Switch the entire system to Licorice (sticks)
view.whole.set_representation(representation='licorice')
```

```python
# Make the whole system semi-transparent
view.whole.set_representation(representation='spacefill', alpha=0.3)
```

## 3. Surgical Control with Regions

While `whole` is useful for the big picture, **Regions** allow us to isolate and style specific subsets of atoms. A region acts as a named selection with its own visual properties.

```python
# Create a region for the first 10 residues
core = view.new_region(selection='1<=group_id<10', tag='nucleo')

# Apply a specific style to this region
core.set_representation(representation='licorice')
```

## 4. Managing Visibility

Both the `whole` system and individual `regions` can be shown or hidden independently. This is essential for focusing on specific interactions.

```python
# Hide the global baseline to see only our 'nucleo' region
view.whole.hide()

# Show it again
# view.whole.show()
```

```python
# You can also access regions by their tag from the viewer registry
view.regions['nucleo'].hide()
```
