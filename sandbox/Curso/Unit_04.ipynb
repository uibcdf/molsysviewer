# Unit 04: Additive Composition

Combining multiple distinct molecular systems into a single viewport.

### Learning Objectives
1. Load multiple coordinate files into the same session using Additive Mode.
2. Inspect loaded block records in the system ledger.
3. Understand how separate, automatically isolated visual regions are created.

## 1. Setting the Stage

Let's initialize a fresh viewer.


```python
import molsysviewer as msv

view = msv.MolSysView()
view
```

## 2. Loading the First System

We first load Crambin.


```python
view.load('1CRN', label='crambin')
```

## 3. Loading the Second System (Additive Mode)

Now, we load a DNA dodecamer (`1BNA`) alongside the protein. Using `mode='add'` ensures both systems coexist.


```python
# Load DNA in additive mode
view.load('1BNA', label='dna', mode='add')
```

## 4. Auditing Multi-System Regions

When we load a second structure, the viewer back-fills the first structure as an automatic region and creates a new region for the second one.


```python
# Inspect active regions
print("Active regions:", list(view.regions.keys()))
```

```python
# Style the DNA region as an orange surface
view.regions['dna'].set_representation(representation='surface', color='orange')
```
