# Unit 07: The Selection Registry

Registering and manipulating named structural subsets.

### Learning Objectives
1. Store selections under named tags in the selection registry.
2. Focus camera snapshots on named selections.
3. Convert selections directly to visual regions.

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Registering Selections

We can save selection queries into the `view.selections` manager.


```python
# Add a named selection called 'active_pocket'
view.selections.add_selection(tag='active_pocket', selection='all within 5.0 angstroms of group_index==15')

# List registered tags (accessed via property)
print("Saved selections:", view.selections.tags)
```

## 3. Operations on Registry Selections

Once registered, we can center the camera or instantiate regions directly.


```python
# Focus camera on the selection
view.selections['active_pocket'].focus()
```

```python
# Convert to a licorice representation region
view.selections['active_pocket'].new_region(tag='pocket_view', representation='licorice')

# Hide global baseline to see the pocket
view.whole.hide()
```
