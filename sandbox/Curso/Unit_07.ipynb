# Unit 07: Labeling and Documenting

Molecular visualization is often about storytelling. To guide an audience through a structure, we need to add textual cues. In this unit, we will learn how to enrich our 3D scene with persistent labels anchored to specific biological features, using both the interactive UI and the Python API.

### Learning Objectives
1. Add text labels interactively using the mouse.
2. Place labels programmatically using MolSysMT selections.
3. Manage the annotation registry (visibility, tags, and deletion).

## 1. Setting the Stage

Let's load our reference protein and prepare the scene.

```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Interactive Labeling (UX)

The easiest way to label a structure is by direct interaction. 

**Instructions:**
1. Click on a residue in the 3D viewer above.
2. Run the cell below to anchor a label to your selection.

```python
# Add a label to the atom/residue you just clicked
label_ux = view.active_selection.add_label(text="Key Residue")
```

## 3. Programmatic Labeling (API)

For reproducible workflows, we can place labels using code. The label's position is automatically calculated as the geometric center of the selection.

```python
# Label a specific residue using MolSysMT syntax
view.annotations.add_annotation(text='Active Site', selection='group_id==20', tag='site_20')

# You can also label specific atoms
view.annotations.add_annotation(text='Atom 100', selection='atom_index==100', tag='tip')
```

## 4. Auditing and Visibility

Labels are registered in the `annotations` manager. We can check their status and toggle visibility using their unique **tags** (not their text content).

```python
# List all active annotation tags
print(f"Active tags: {view.annotations.tags}")

# See detailed metadata for all labels
view.annotations.info()
```

```python
# Hide a specific label by its tag
view.annotations['site_20'].hide()

# Show it again
# view.annotations['site_20'].show()
```

## 5. Deleting and Cleaning

To remove labels from the scene, you can delete them individually or clear the entire registry.

```python
# Remove one specific label
view.annotations['site_20'].delete()

# Or clear all annotations to start fresh
view.annotations.clear()
```

### Summary
Annotations bridge the gap between structural data and human interpretation.
- Use **Interactive (UX)** clicks for quick exploration.
- Use **Programmatic (API)** selections for reproducible figures.
- Always manage annotations via their unique **tags**.
