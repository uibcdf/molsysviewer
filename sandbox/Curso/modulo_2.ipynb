```python
import molsysviewer as msv
```

```python
view = msv.MolSysView()
```

```python
view.load('1CRN', label='crambin')
```

```python
view
```

```python
view.whole.set_representation(representation='licorice')
```

```python
view.whole.set_representation(representation='point')
```

```python
core = view.new_region(selection='1<=group_id<10', tag='nucleo')
```

```python
core.set_representation(representation='cartoon')
```
