# Unit 08: Whole vs Regions

Segmenting the structure into visual units.

### Learning Objectives
1. Control global scene background with the Whole manager.
2. Create custom named Regions to isolate specific structural regions.
3. Isolate chains or molecules using automated segmentation.

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Global Control vs. Local Regions

`view.whole` controls the entire system, while `view.regions` controls sub-selections.


```python
# Hide the entire protein representation
view.whole.hide()
```

```python
# Create a region for the core residues 1 to 10
core_region = view.new_region(selection='group_id<10', tag='core')

# Display only the core region
core_region.show()
core_region.focus()
```

## 3. Automated Segmentation

We can segment the entire molecular structure in one call using topological elements.


```python
# Segment the system by chain
view.make_regions_by(element='chain')
print("Current regions:", list(view.regions.keys()))
```
