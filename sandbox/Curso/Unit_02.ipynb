# Unit 02: Advanced Selection Registry

Analysis often involves identifying specific structural subsets, such as a binding pocket or a catalytic triad. In this unit, we will learn how to use the **Selection Registry** to save these complex queries, reuse them across the session, and convert them into visual objects like Regions.

### Learning Objectives
1. Save interactive selections with semantic tags.
2. Master MolSysMT selection syntax (`within`, `backbone`, `sidechain`).
3. Convert persistent selections into **Regions** and **Annotations**.

## 1. Setting the Stage

Let's load a protein-ligand complex to perform advanced queries.

```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Saving an Interactive Selection

**Instructions:**
1. Click on any residue in the viewer above.
2. Run the cell below to "freeze" that selection into the registry.

```python
print(f"¿Hay algo seleccionado?: {not view.active_selection.is_empty()}")
```

```python
# Save your click as a named selection
my_pick = view.active_selection.save(tag='favorite_residue')

print(f"Stored selections: {view.selections.tags()}")
```

## 3. Mastering Selection Syntax

You can define selections programmatically using the full power of MolSysMT. This allows for scientific queries based on distance, chemical properties, or secondary structure.

```python
# Create a selection for atoms within 5.0 Angstroms of residue 20
pocket = view.selections.add_selection(tag='pocket', selection='all within 5.0 angstroms of group_index==10')
```

```python
# Check how many atoms were captured
print(f"Atoms in pocket: {len(pocket.atom_indices)}")
```

## 4. Converting Selections to Visual Objects

Once a selection is saved, it becomes a versatile template. You can center the camera on it or transform it into a visual Region with its own style.

```python
view.selections.tags
```

```python
# Focus the camera on the pocket
view.selections['pocket'].focus()
```

```python
# Create a high-detail region from this selection
view.selections['pocket'].new_region(tag='visual_pocket', representation='licorice')
```

```python
# Hide the rest of the system to inspect the pocket
view.whole.hide()
```

### Summary
The Selection Registry is the bridge between analysis and visualization.
- Save interactive work via `active_selection.save()`.
- Build complex queries with MolSysMT syntax.
- Use saved selections to drive regions and camera focus.
