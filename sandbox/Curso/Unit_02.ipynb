# Unit 02: First Load & Metadata

In computational structural biology, the very first step of any visual study is loading structural data. A file is not just a collection of 3D coordinates; it contains metadata like chain names, residue IDs, and chemical elements.

In this unit, we will learn how to load a structure, label it for identification, and query its basic composition using the viewer's information registry.

### Learning Objectives
1. Load a molecular system into the viewer using standard PDB IDs.
2. Customize load identifiers with the `label` parameter.
3. Inspect load records using `view.load_blocks`.
4. Query the loaded structure's metadata using the `view.info()` command.

## 1. Setting the Stage: Launching the Canvas

As before, we import the library and instantiate our default molecular workbench.

```python
import molsysviewer as msv

# Create the viewer instance
view = msv.MolSysView()

# Display the widget
view
```

## 2. Interactive UX Action: Loading and the Canvas UI

When you run the next cell, keep your eyes on the 3D widget above. You will see:
1.  A loading bar at the top or bottom of the widget indicating downloading and parsing.
2.  The structure of Crambin (`1CRN`) rendering in the center of the canvas.
3.  The side navigate panel update to reflect the newly loaded component.

```python
# Load Crambin (PDB ID: 1CRN) with an explicit label
view.load('1CRN', label='crambin')
```

```python
view.load('2LAO', label='xxx')
```

## 3. Programmatic API Equivalent: Querying Load Blocks

To keep track of what has been loaded programmatically (vital for reproducibility), the viewer keeps a read-only ledger of load blocks. Let's inspect the `view.load_blocks` property.

```python
view.load_blocks
```

```python
# Inspect the load blocks registry
blocks = view.load_blocks
print("Loaded blocks count:", len(blocks))
print("Block 1 details:", blocks[0])
```

## 4. Audit & Diagnostics: Inspecting Molecular Metadata

MolSysViewer provides a built-in `info()` method to query structural details. By default, it returns a dual-section info record (`ViewerInfo`) summarizing:
*   **MolSys Section:** The underlying molecular system (number of atoms, groups, chains, etc.).
*   **View Section:** Visual status (active representations, colors, visibility).

Let's query this dual-context metadata.

```python
# Retrieve and display the combined info table
view.info()
```

You can focus the query on a specific context using the `source` parameter:
*   `source="molsys"`: Only shows structural/chemical data from the backend.
*   `source="view"`: Only shows representation and styling data.

```python
# Audit only the backend molecular data
view.info(source='molsys')
```

## 5. Key Takeaway

You have loaded your first structure and audited its content programmatically!

*   **Labels are Identifiers:** The `label` parameter is the name that links the physical file to the visual object.
*   **Auditing State:** Properties like `view.load_blocks` and methods like `view.info()` are essential for pipeline logging and verification.
