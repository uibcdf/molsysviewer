## 1. Setting the Stage: Initializing the Viewer

We start by importing `molsysviewer` and initializing a `MolSysView` instance. If a `MolSysView` instance is the last expression in a notebook cell, Jupyter will render the interactive 3D canvas automatically.

```python
import molsysviewer as msv

# Create the default viewer instance
view = msv.MolSysView()

# Render the widget
view
```

## 2. Interactive UX Action: Personalizing Canvas Controls

By default, the viewer starts in `classic` controls mode, displaying text buttons at the top of the canvas (*Reset*, *Full*, *Bg*, *Spin*, *Swing*, *Pop*).

MolSysViewer also offers a **Minimal Controls Mode**, which hides the text buttons and groups controls into a clean, modern SVG icon cluster (panel, fullscreen, popup) plus a help button `?`. Under this mode, scene actions (like resetting the camera or spinning the model) are moved to the empty-canvas context menu (right-click on the background).

Let's test this layout style interactively. Try right-clicking the canvas background in the viewer above or pressing the `H` key on your keyboard to toggle the help overlay.

## 3. Programmatic API Equivalent: Configuring Layouts via Python

Instead of relying on mouse clicks, we can programmatically set these preferences at initialization.

`MolSysView` accepts two experimental arguments to customize its look:
*   `controls_mode`: `"classic"` (default buttons) or `"minimal"` (SVG cluster, background context-menu).
*   `panel_mode_style`: `"drawer"` (panels slide in and push the viewport, default) or `"floating"` (panels overlay on top of the canvas, card style).

Let's instantiate a second viewer using **Minimal controls** and a **Floating layout** card style.

```python
# Create a minimalist viewer with a floating card layout
view_mini = msv.MolSysView(controls_mode="minimal", panel_mode_style="floating")

# Render the minimalist viewer
view_mini
```

## 4. Controlling Panels and Workspaces

MolSysViewer has a dual-panel setup:
*   **Navigate Panel (`navigate`):** Left-side panel for structure loading, selections, and topology exploration (keyboard shortcut: `H` or `N`).
*   **Workbench Panel (`workbench`):** Right-side panel for representations, colors, shapes, measurements, and annotations (keyboard shortcut: `W`).

We can programmatically open, collapse, and switch between these panels using the `set_panel_mode()` method. Let's practice with the first viewer.

```python
# Open the Navigate panel on our first viewer
view.set_panel_mode(panel="navigate")
```

```python
# Switch to the Workbench panel, expanding it if collapsed
view.set_panel_mode(panel="workbench")
```

```python
# Collapse all active panels to maximize 3D viewport area
view.set_panel_mode(expanded=False)
```

## 5. Audit & Diagnostics: Inspecting Layout State

In MolSysViewer, the Python kernel and the JS frontend run in sync. You can query the current layout state of the panels at any time using the `get_panel_mode_state()` method.

```python
# Inspect the current panel and workspace state reported by the frontend
layout_state = view.get_panel_mode_state()
print("Current Layout State:", layout_state)
```
