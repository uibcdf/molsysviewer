# Unit 01: Loading and Identity

In this first unit, we will learn how to bring molecular structures into the virtual world of **MolSysViewer**. We will explore the core concepts of structure loading, naming (labels), and how the viewer manages multiple systems simultaneously.

### Learning Objectives
1. Initialize the viewer widget in a Jupyter environment.
2. Load molecular structures using PDB identifiers.
3. Master **Additive Loading** to build multi-component scenes.
4. Audit the viewer's internal state using the `info()` method.

## 1. Initializing the Viewer

First, we need to import the library and create an instance of the `MolSysView` class. This object acts as the orchestrator for our 3D scene.

```python
import molsysviewer as msv

# Create the viewer instance
view = msv.MolSysView()

# Display the widget
view
```

## 2. Loading your first Structure

We use the `load()` method to fetch structures. By providing a `label`, we give our structure a semantic name that will be used later for organization and styling.

```python
# Load Crambin (1CRN) with a descriptive label
view.load('1CRN', label='crambin')
```

Many molecular systems include information about their simulation box or unit cell. We can visualize these boundaries using the `show_box()` method.

```python
# Display the unit cell / simulation box
view.show_box(color='grey', alpha=0.5)
```

```python
view.hide_box()
```

## 3. Additive Loading (The Power of `add`)

MolSysViewer is intelligent: when you call `load()` again, it detects that a system is already present and offers to **add** the new structure instead of replacing it. This automatically organizes each load block into its own **Region**.

```python
# Load a DNA dodecamer (1BNA) alongside the protein
view.load('1BNA', label='dna', mode='add')
```

## 4. Auditing the Scene

To understand what is happening "under the hood", we can use the `info()` method. This provides a comprehensive summary of both the **Molecular System** (atoms, residues) and the **Viewer State** (regions, active selection).

```python
# Inspect the dual-context state of the session
view.info()
```

Every time you successfully load a structure, the viewer records it in the `load_blocks` registry. This tracks the atom ranges and labels for every component of your scene.

```python
# Check the accounting of loaded systems
view.load_blocks
```

Notice how the viewer automatically created regions named after our labels (`crambin` and `dna`). These regions allow us to style each structural block independently in future units.

```python
# List the automatically created regions
list(view.regions.keys())
```

```python
view.regions['dna'].info()
```

```python
view.regions['dna'].set_representation(representation='surface', color='orange')
```
