# Unit 22: The Backend Engine

Accessing the underlying MolSysMT data structure.

### Learning Objectives
1. Access the molsys backend property.
2. Perform analytical queries via MolSysMT.
3. Feed analysis data back into regions.

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv
import molsysmt as msm

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Accessing the Backend System

`view.molsys` provides direct access to the underlying coordinate and topology data structure.


```python
# Inspect system data using molsysmt
print("Backend molecular system type:", type(view.molsys))
msm.info(view.molsys)
```

## 3. Querying with MolSysMT

We can find Tyr residues and create regions from backend data.


```python
# Select atom indices via backend
tyr_indices = msm.select(view.molsys, selection='res_name=="TYR"')

# Feed indices to create region
view.new_region(atom_indices=tyr_indices, tag='tyrosines', representation='licorice')
```
