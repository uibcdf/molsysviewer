# Unit 23: Interaction Forensics (Custom Tools)

MolSysViewer provides high-level tools for analysis, but sometimes you need to build your own. In this unit, we will learn how to access the **Raw Interaction Events** to capture exactly where and what the user is touching. This transforms the viewer into a sensory input device for your Python scripts.

### Learning Objectives
1. Capture raw `click` and `hover` events from the frontend.
2. Inspect event metadata (Cartesian coordinates, atom indices, object types).
3. Trigger custom Python logic based on user input.

## 1. Setting the Stage

Let's load a system and prepare to "listen" to its events.

```python
import molsysviewer as msv
import numpy as np

view = msv.MolSysView()
view.load('1CRN')
view

print("Instructions: Click anywhere in the viewer above, then run the cells below.")
```

## 2. Inspecting the Last Click

The `get_last_click_event()` method returns a dictionary containing everything the frontend knows about the last user interaction.

```python
# Retrieve the raw event data
event = view.get_last_click_event()

if event:
    print(f"Object Kind: {event.get('kind')}")
    print(f"Cartesian Coordinates (nm): {event.get('cartesian_position')}")
    print(f"Atoms under mouse: {event.get('atom_indices')}")
else:
    print("No click recorded yet.")
```

## 3. Building a Custom Tool

We can use this metadata to create a custom tool. For example, let's write a function that calculates the distance from the clicked point to the center of mass of the entire system.

```python
import pyunitwizard as puw

def custom_distance_tool():
    event = view.get_last_click_event()
    if not event or 'cartesian_position' not in event:
        print("Please click on the structure first.")
        return
    
    pos = puw.quantity(event['cartesian_position'], 'nm')
    origin = puw.quantity([0, 0, 0], 'nm')
    
    # Compute distance using NumPy
    dist = np.linalg.norm(puw.get_value(pos - origin))
    print(f"Distance from Click to Origin: {dist:.3f} nm")

custom_distance_tool()
```

## 4. Hover Events (Dynamic Insight)

While clicking is a deliberate action, **Hovering** provides continuous feedback. You can use `get_last_hover_event()` to get information about what the user is currently looking at, without requiring a click.

```python
# Get hover info
hover_data = view.get_last_hover_event()
if hover_data:
    print(f"Currently hovering over: {hover_data.get('kind')}")
```

### Summary
Interaction Forensics allows you to expand MolSysViewer's logic with your own Python code.
- Capture user input via `get_last_click_event()`.
- Use Cartesian and atomic metadata to drive calculations.
- Bridge the gap between visual exploration and custom biophysical analysis.
