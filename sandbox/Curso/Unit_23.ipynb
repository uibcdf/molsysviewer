# Unit 23: Coordinate Engineering

Modifying atom coordinates dynamically.

### Learning Objectives
1. Extract coordinate arrays.
2. Apply spatial perturbations (translations).
3. Push coordinates back to the viewer.

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv
import pyunitwizard as puw

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Extracting Coordinates

We can get coordinates in nanometers (nm).


```python
# Get coordinate array
coords = view.get_coordinates(selection='all')
print("Coordinate shape:", coords.shape)
```

## 3. Perturbing Coordinates

Let's apply a translation vector and push coordinates back to update the visualization.


```python
# Shift coordinates of residue 10
indices = view.select('res_id 10')
res_coords = view.get_coordinates(selection=indices)
new_coords = res_coords + puw.quantity([0.5, 0.0, 0.0], 'nm')

# Push back coordinates
view.set_coordinates(new_coords, selection=indices)
```
