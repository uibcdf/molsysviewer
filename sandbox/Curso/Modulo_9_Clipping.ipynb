Módulo 9: Secciones y Planos de Corte (Visualización Avanzada)

  En este capítulo, aprenderemos a controlar los planos de corte (Z-Clipping) para "rebanar" el sistema y ver qué ocurre en el corazón de la proteína.

  Objetivos Pedagógicos:
   1. Ajustar el plano de corte frontal (near).
   2. Controlar la visibilidad del plano lejano (far).
   3. Entender el uso de planos de corte para exploración interna.


```python
import molsysviewer as msv
view = msv.MolSysView()
view.load('1CRN')
```

```python
view
```

```python
# El argumento 'near' acepta un porcentaje (0-100)
# Vamos a cortar el primer 30% del frente de la escena
view.scene.set_clip_planes(near=70.0)
```

```python
view.scene.set_clip_planes(far=False, near=80.0)
```

```python
# Aprovechamos para probar la nueva iluminación que también ha aparecido
view.scene.set_lighting(ambient=0.8, diffuse=0.3)
```
