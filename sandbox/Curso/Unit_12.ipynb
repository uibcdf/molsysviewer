# Unit 12: Trajectories & Structure Navigation

Molecular systems are rarely static. Whether you are analyzing a Molecular Dynamics (MD) trajectory or comparing docking poses, you need to navigate through frames. In this unit, we will use a real MD snippet to master the **Player Manager** and control the structural timeline.

### Learning Objectives
1. Load a trajectory from the **Demo Catalog**.
2. Navigate through structures using the `view.player` manager.
3. Control playback speed, direction, and looping.
4. Query the current structural state programmatically.

## 1. Setting the Stage

We will load a pentalanine trajectory from the built-in demo catalog. This system contains 200 frames of a small peptide in motion.

```python
import molsysviewer as msv

# Load a real trajectory demo
view = msv.demo['pentalanine']

# Display the viewer with its trajectory player
view
```

## 2. Programmatic Navigation

The `view.player` manager allows you to jump to specific points in the trajectory. This is vital for linking structural visualization to analytical events (e.g., "Jump to the folding event").

```python
# Jump to the 50th frame (index 49)
view.player.go_to_structure(index=49)

# Inspect the current state
print(f"Current frame index: {view.player.index}")
print(f"Total frames available: {view.player.n_structures}")
```

## 3. Controlling the Animation

You can start, pause, and adjust the playback speed (FPS) directly from Python. You can also change the direction or set the player to 'loop' or 'ping-pong' mode.

```python
# Play the trajectory at 24 frames per second
view.player.play(fps=24)

# Change direction to backward
view.player.set_direction('backward')

# Pause the animation
# view.player.pause()
```

## 4. Frame-by-Frame Inspection

For detailed analysis, use the stepping methods. `step_forward()` and `step_backward()` allow you to advance the system by a specific number of frames.

```python
# Advance by 5 frames
view.player.step_forward(n=5)

print(f"New frame: {view.player.index}")
```

### Summary
The Player Manager makes molecular dynamics accessible and reproducible.
- Load trajectories directly via `msv.demo` or standard `load()`.
- Navigate semantically with `go_to_structure()`.
- Command the animation loop with `play()` and `set_fps()`.
