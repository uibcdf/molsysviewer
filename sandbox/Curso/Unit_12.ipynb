# Unit 12: Quantitative Color Mapping (Heatmaps)

Mapping numerical arrays onto physical residues using colormaps.

### Learning Objectives
1. Construct scalar values (conservation, dynamics).
2. Map continuous values onto atom or residue selections.
3. Reset color schemes back to standard baselines.

## 1. Setting the Stage

Load Crambin. We import NumPy to construct value profiles.


```python
import molsysviewer as msv
import numpy as np

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Mapping continuous values

Crambin has 46 residues. Let's create an array of fake scores from 0.0 to 1.0.


```python
# Create 46 linear scores
scores = np.linspace(0.0, 1.0, 46)

# Map values to residues (groups) using viridis colormap
view.whole.set_color_by_values(values=scores, element='group', palette='viridis')
```

## 3. Reverting Colors

To clear continuous overrides, use `reset_colors()`.


```python
# Clear continuous mapping
view.whole.reset_colors()
```
