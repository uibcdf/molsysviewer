# Unit 27: Professional Shareables (Figures & HTML)

Exporting high-quality figures and standalone sessions.

### Learning Objectives
1. Export snapshots as high-resolution PNGs.
2. Create reusable FigureSpec recipes.
3. Save sessions as standalone interactive HTML files.

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv
from molsysviewer.figures import FigureSpec

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. FigureSpec Recipes & Image Exports

We can define high-quality publication layout specs.


```python
# Define a light publication theme spec
pub_spec = FigureSpec(background='white', scale=3.0)

# Export the image using the spec
view.export.image('my_crambin_fig.png', figure_spec=pub_spec)
```

## 3. Exporting Standalone Interactive HTML

To share the interactive 3D model with colleagues (opens in any browser without Python):


```python
# Save the session to an HTML file
view.export.html('crambin_session.html', title='Crambin Analysis', mode='standalone')
```
