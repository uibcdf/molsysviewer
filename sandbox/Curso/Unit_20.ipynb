# Unit 20: The Computational Engine (MolSysMT Integration)

MolSysViewer is not a standalone tool; it is a visual extension of the **MolSysMT** ecosystem. In this unit, we will learn how to access the underlying molecular system and use MolSysMT's analytical power to drive our visualizations.

### Learning Objectives
1. Access the raw molecular data via the `molsys` property.
2. Perform structural analysis (e.g., counting atoms, identifying residues).
3. Use MolSysMT results to update the viewer state programmatically.

## 1. Accessing the Data Engine

When you load a structure, MolSysViewer creates a MolSysMT object internally. You can access it using the `view.molsys` property.

```python
import molsysviewer as msv
import molsysmt as msm

view = msv.MolSysView()
view.load('1CRN', label='crambin')

# view.molsys provides direct access to the data
print(f"Underlying system type: {type(view.molsys)}")
msm.info(view.molsys)
```

## 2. Dynamic Analysis

Because `view.molsys` is a live MolSysMT object, you can run any analytical function on it. Let's find all Tyrosine residues and use that information to create a region.

```python
# Use MolSysMT to find specific atom indices
tyr_atoms = msm.select(view.molsys, selection='res_name == "TYR"')

print(f"Atoms belonging to Tyrosines: {tyr_atoms}")

# Use the result to create a region in the viewer
view.new_region(atom_indices=tyr_atoms, tag='tyrosines', representation='licorice')
view
```

## 3. Modifying the System

Advanced users can even modify the coordinates or topology via MolSysMT and see the results in the viewer. **Note:** Direct modifications require a view rebuild or coordinate update to be visible.

```python
# Example: Shift the coordinates using MolSysMT
msm.basic.translate(view.molsys, translation=[0.5, 0.5, 0.5])

# Update the viewer to reflect the new coordinates
view.set_coordinates(msm.get(view.molsys, element='atom', coordinates=True))
```

### Summary
The integration with MolSysMT transforms the viewer into an analytical workstation.
- Access the data model via `view.molsys`.
- Combine MolSysMT analysis with visual feedback.
- Build complex, data-driven scenes programmatically.
