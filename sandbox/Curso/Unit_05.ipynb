# Unit 05: Interactive Selections (UX)

Capturing and reacting to mouse click events on the canvas.

### Learning Objectives
1. Inspect the selection status in vivo using the active selection interface.
2. Clear selections programmatically.
3. Retrieve detailed mouse interaction event structures from the canvas.

## 1. Setting the Stage

Let's load Crambin.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Interactive UX Action: Click on Atoms

Try clicking on a few residues or atoms in the Crambin structure above. Notice how they become highlighted in the canvas.


## 3. Programmatic API Equivalent: Auditing Selection Status

We can check if anything is currently selected and clear it programmatically.


```python
# Check selection status
print("Is active selection empty?:", view.active_selection.is_empty())
```

```python
# Clear the active selection
view.active_selection.clear()
```

## 4. Retrieving the Interaction Event

We can fetch the exact details of the last click event, including atom indices and Cartesian coordinates.


```python
# Get last active selection event
event = view.get_last_active_selection_event()
print("Event Details:", event)
```
