el alumno aprenderá que una proteína no es estática, sino que tiene dinámica.

  Objetivos Pedagógicos:
   1. Cargar un sistema con múltiples estructuras (frames).
   2. Usar la barra de reproducción del visor.
   3. Añadir estructuras dinámicamente.

```python
import molsysviewer as msv
import molsysmt as msm

# 1. Cargamos una proteína con varios frames (usamos Crambin como base)
view = msv.MolSysView()
view.load('1CRN', label='dinamica')
```

```python
molsys2 = view.molsys.copy()
msm.structure.translate(molsys2, '[0.5, 0.5, 0.5] nm', in_place=True)
```

```python
view.append_structures(molsys2)
```
