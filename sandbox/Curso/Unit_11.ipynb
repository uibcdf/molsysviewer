# Unit 11: Categorical Color Schemes

Coloring structures based on discrete biological properties.

### Learning Objectives
1. Color elements by standard CPK scheme.
2. Color chains by distinct identifiers.
3. Color structural folds by secondary structure (helices, sheets).

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Coloring by Element

By default, atoms are colored by standard CPK elements.


```python
# Revert/set to element color scheme
view.whole.set_representation(color_scheme='element')
```

## 3. Coloring by Secondary Structure

We can highlight alpha-helices, beta-sheets, and loops.


```python
# Color by secondary structure folds
view.whole.set_representation(color_scheme='secondary-structure')
```
