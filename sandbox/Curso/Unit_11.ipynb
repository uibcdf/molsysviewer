# Unit 11: Color Schemes and Palettes

Color is a powerful dimension in molecular visualization. It can be used to identify components, highlight chemical properties, or visualize complex numerical data like atomic charges or conservation scores. In this unit, we will learn how to master the **Color Registry** and apply dynamic gradients to our models.

### Learning Objectives
1. Use built-in categorical color schemes (`element`, `chain`).
2. Apply quantitative gradients using **Scalar Value Mapping**.
3. Integrate **Matplotlib** colormaps directly into the 3D canvas.
4. Manage colors per-atom or per-residue.

## 1. Setting the Stage

Let's load a protein structure. By default, it uses the standard 'element' color scheme.

```python
import molsysviewer as msv
import numpy as np

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Categorical Schemes

You can change the coloring logic of any representation using the `color_scheme` parameter. Standard schemes include `element` (CPK), `chain`, and `secondary-structure`.

```python
# Color by secondary structure (Helices in red, Sheets in blue)
view.whole.set_representation(color_scheme='secondary-structure')
```

## 3. Quantitative Coloring (Heatmaps)

One of the most powerful features of **MolSysViewer** is the ability to map numeric arrays directly to the canvas. The `set_color_by_values()` method automatically normalizes your data and applies a colormap.

```python
# Imagine we have a list of 'Importance' scores for every residue (group)
n_residues = 46 # Crambin has 46 residues
scores = np.linspace(0, 1, n_residues)

# Apply the scores using a Matplotlib palette
view.whole.set_color_by_values(
    values=scores, 
    element='group', 
    palette='viridis'
)
```

## 4. Per-Atom Precision

You can also provide data at the atomic level. This is ideal for visualizing B-factors or partial charges directly from your calculation results.

```python
# Color every atom randomly to test the mapping throughput
random_data = np.random.rand(view.molsys.get_n_atoms())

view.whole.set_color_by_values(
    values=random_data, 
    element='atom', 
    palette='plasma'
)
```

To revert all custom colors and return to the standard representation theme, use `reset_colors()`.

```python
# Revert to the default 'element' colors
view.whole.reset_colors()
```

### Summary
Colors transform a 3D model into a data-rich analytical tool.
- Use `color_scheme` for structural identification.
- Use `set_color_by_values()` for quantitative mapping.
- Integrate Matplotlib palettes seamlessly.
