# Unit 17: Dynamic Layering

Organizing scenes into visibility categories.

### Learning Objectives
1. Assign tags to layer groups.
2. Group regions, shapes, and annotations under a single layer.
3. Toggle visibility of layers in mass.

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Grouping Visual Objects under a Layer

We can tag diverse visual objects (regions, shapes) with a shared `layer_tag`.


```python
# Create region under 'catalytic_site' layer
view.new_region(selection='group_index==10', tag='active_res', layer_tag='catalytic_site')

# Create sphere marker under the same layer
view.shapes.add_sphere(center='[1,1,1] nm', radius='0.5 nm', tag='marker', layer_tag='catalytic_site')
```

## 3. Toggling Layer Visibility

We can show or hide the entire layer, toggling all member visibilities at once.


```python
# Hide the entire layer
view.layers['catalytic_site'].hide()
```

```python
# Show it again
view.layers['catalytic_site'].show()
```
