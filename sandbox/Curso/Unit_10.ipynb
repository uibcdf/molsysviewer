## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Built-in Mol* Presets

Presets apply multiple representation styles and themes at once.


```python
# Apply high-detail sphere representations
view.whole.set_representation(preset='atomic-detail')
```

```python
# Switch to clean polymer-cartoon style
view.whole.set_representation(preset='polymer-cartoon')
```
