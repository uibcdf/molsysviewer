# Unit 10: Advanced Scene Organization

As your analysis grows, your 3D scene will become crowded with structural regions, geometric markers, and annotations. In this unit, we will learn how to master the **Hierarchy of Layers** to organize complex scenes, group related objects, and perform mass visibility updates.

### Learning Objectives
1. Group disparate objects (Regions and Shapes) into a single logical **Layer**.
2. Move existing objects between layers using `set_layer_tag()`.
3. Perform surgical deletion of individual components within a group.
4. Audit the scene hierarchy using the `view.layers` registry.

## 1. Setting the Stage

Let's load a protein and prepare multiple analytical elements.

```python
import molsysviewer as msv
view = msv.MolSysView()
view.load('1CRN', label='protein')
view
```

## 2. Grouping at Creation

When you create a region or a shape, you can specify a `layer_tag`. If multiple objects share the same `layer_tag`, they can be controlled as a single unit.

```python
# Create an analytical group called 'site_A'
active_res_region = view.new_region(selection='group_id==10', tag='active_res', layer_tag='site_A')
```

```python
active_site = active_res_region.get_center()
```

```python
active_site='[0,0,0] nm'
```

```python
view.shapes.add_sphere(centers=active_site, radii='0.5 nm', tag='marker', layer_tag='site_A')

print(f"Layer 'site_A' members: {list(view.layers['site_A'].members.keys())}")
```

## 3. Dynamic Regrouping

You don't always know the hierarchy at the time of creation. You can move any existing object into a different layer using the `set_layer_tag()` method.

```python
# Create an annotation independently
note = view.annotations.add_annotation(text='Important point', selection='group_id==10', tag='note1')

# Now move it into the 'site_A' group
view.annotations['note1'].set_layer_tag('site_A')

print(f"Updated 'site_A' members: {list(view.layers['site_A'].members.keys())}")
```

## 4. Mass Control and Surgical Deletion

Using the layer registry, you can show/hide dozens of objects with a single command. You can also delete a specific shape from a layer without affecting the rest of the group.

```python
# Hide everything related to 'site_A'
view.layers['site_A'].hide()
```

```python
# Delete only the sphere marker, keep the region and label
view.shapes['marker'].delete()
```

```python
# Show the remaining members
view.layers['site_A'].show()
```

### Summary
Advanced Scene Organization transforms a collection of objects into a structured scientific project.
- Use `layer_tag` to define visibility groups.
- Use `set_layer_tag()` to regroup objects dynamically.
- Access group members via `view.layers['tag'].members`.
