# Unit 16: Scene Merging & Composition

In complex projects, you might work on different components of a system separately. **MolSysViewer** allows you to fuse independent viewer sessions into a single, unified scene. This unit explores the `merge()` tool, which intelligently combines molecular data, visual styles, and annotations from multiple sources.

### Learning Objectives
1. Combine multiple independent `MolSysView` instances.
2. Understand how atom indices and tags are remapped automatically.
3. Learn how global scene properties (camera, background) are inherited during a merge.

## 1. Preparing the Source Views

Let's create two separate views with different contents and styles to simulate a collaborative or modular workflow.

```python
import molsysviewer as msv

# View 1: Protein with a specific region
view1 = msv.MolSysView()
view1.load('1CRN', label='protein')
view1.new_region(selection='res_id 1:10', tag='core', representation='licorice')

# View 2: DNA with a measurement
view2 = msv.MolSysView()
view2.load('1BNA', label='dna')
view2.whole.set_representation(representation='licorice')
view2.measurements.add_distance(atom_indices_a=[0], atom_indices_b=[10], tag='dna_dist')
```

## 2. Performing the Merge

The `msv.tools.basic.merge()` function takes a list of views and returns a new, independent viewer containing the sum of all parts.

```python
from molsysviewer.tools.basic import merge

# Fuse the protein and DNA views
combined_view = merge([view1, view2])

# Display the result
combined_view
```

## 3. Auditing the Combined State

Notice that the `combined_view` has inherited the regions from `view1` and the measurements from `view2`. MolSysViewer has automatically updated the atom indices to ensure everything stays in the right place.

```python
print(f"All regions: {list(combined_view.regions.keys())}")
print(f"All measurements: {list(combined_view.measurements.tags())}")

# If tags had collided (e.g., both had a 'core' region), 
# the second one would be renamed to 'core__2' automatically.
```

## 4. Inheriting Scene Properties

The first view in the list (`view1`) acts as the "Host". The background color, lighting settings, and the last camera snapshot of the merged view will be taken from this primary source.

### Summary
Scene merging allows for modular visual design and collaboration.
- Use `merge()` to combine independent analysis sessions.
- Regions, Shapes, and Measurements are preserved and remapped.
- Conflicting tags are resolved deterministically.
