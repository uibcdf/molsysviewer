# Unit 16: Abstract Geometric Shapes

Drawing non-atomic geometric shapes on the scene.

### Learning Objectives
1. Add spheres at Cartesian coordinates.
2. Draw links/cylinders between points.
3. Manipulate coordinate vectors programmatically.

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Drawing Spheres

We can add spheres using center points in nanometers (nm).


```python
# Add a green semi-transparent sphere at the origin
sph = view.shapes.add_sphere(center='[0, 0, 0] nm', radius='1.0 nm', color=0x00FF00, alpha=0.5, tag='origin_sph')

# Focus camera on the shape
view.shapes['origin_sph'].focus()
```

## 3. Drawing Links (Cylinders)

We can draw lines between two coordinates.


```python
# Draw a connecting cylinder
view.shapes.add_links(coordinate_pairs=[[[0,0,0], [1,1,1]]], tag='link_1')
```
