# Unit 24: Topological Rebuilding (Editions)

Applying system modifications and reconciling state indices.

### Learning Objectives
1. Apply structural mutations/deletions on backend.
2. Update viewer systems using apply_system_edit.
3. Verify annotation and measurement index survival.

## 1. Setting the Stage

Load Crambin. Add a label to atom index 10.


```python
import molsysviewer as msv
import molsysmt as msm

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view.annotations.add_annotation(text='Target Atom', selection='atom_index==10', tag='target')
view
```

## 2. Applying Topological Edits

When we delete atoms, indices shift. We must notify the viewer to reconcile annotations.


```python
# Remove first 5 atoms from backend molsys
new_molsys = msm.basic.remove(view.molsys, selection='atom_index<5')

# Define the index mapping map (old -> new)
# (For this example, let's notify the viewer of system edits)
view.apply_system_edit(new_molsys)
```
