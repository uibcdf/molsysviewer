  Objetivos Pedagógicos:
   1. Capturar clics del ratón en Python.
   2. Diferenciar entre hover (pasar por encima) y click (seleccionar).

```python
import molsysviewer as msv
view = msv.MolSysView()
view.load('1TCD') #1CRN
view
```
