  Objetivos Pedagógicos:
   1. Añadir formas básicas (esferas) a la escena.
   2. Entender que las formas viven en coordenadas cartesianas independientes de la molécula.
   3. Aprender a añadir múltiples formas de manera eficiente (Batch).
   4. Organizar formas mediante tags y capas.


```python
import molsysviewer as msv
import numpy as np
view = msv.MolSysView()
view.load('1CRN', label='proteina')
```

```python
esfera = view.shapes.add_sphere(center='[0.0, 0.0, 0.0] nm', radius='1.0 nm', color=0xFF0000, 
                                alpha=0.4, tag='centro_de_masas')
```

```python
import pyunitwizard as puw
centros = puw.quantity([[2.0, 0.0, 0.0],[0.0, 2.0, 0.0],[0.0, 0.0, 2.0]], 'nm')
```

```python
enjambre = view.shapes.add_spheres(
    centers=centros, radii=puw.quantity(0.3,'nm'),
    colors=0x00FF00,
    alpha=0.6,
    tag='puntos_de_control'
)
```

```python
print(f"Capas creadas: {list(view.layers.keys())}")
```

```python
print(f"Shapes creadas: {list(view.shapes.keys())}")
```
