Etiquetado y Documentación (Anotaciones)

  Objetivos Pedagógicos:
   1. Añadir etiquetas de texto (labels) ancladas a átomos.
   2. Crear anotaciones desde una selección activa.
   3. Gestionar el inventario de anotaciones (ocultar/borrar).


```python
import molsysviewer as msv
view = msv.MolSysView()
view.load('1CRN')
view
```

--- INSTRUCCIONES PARA EL ALUMNO ---
1. Haz clic en un residuo que quieras etiquetar.
2. Ejecuta la celda de abajo.

```python
label = view.active_selection.add_label(text="Residuo Clave")
```

```python
print(f"Anotaciones activas: {view.annotations.info()}")
```

```python
label.hide()
```

```python
label.delete()
#view.annotations.delete('annotation1')
```

```python
print(f"Anotaciones activas: {view.annotations.info()}")
```

```python
view.annotations.add_label(text="Sitio 15", group_index=[15], tag="mi_marca")
```

```python
view.select(selection='atom_index==100')
view.active_selection.add_label(text="Átomo 100")
```

```python
print(f"Todas las etiquetas: {view.annotations.tags()}")
```

```python
view.annotations.clear()
```

```python
print(f"Todas las etiquetas: {view.annotations.tags()}")
```
