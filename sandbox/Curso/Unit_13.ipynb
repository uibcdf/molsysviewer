# Unit 13: Camera Snapshots & Zoom

Querying and restoring viewpoint snapshots.

### Learning Objectives
1. Focus camera snapshots on selections.
2. Export camera state vectors.
3. Restore viewpoint snapshots dynamically.

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Capturing Current Viewpoint

We can capture the exact camera coordinates using `get_snapshot()`.


```python
# Get snapshot vector
my_camera = view.camera.get_snapshot()
print("Camera parameters:", list(my_camera.keys()))
```

## 3. Zooming and Restoring

Let's zoom into a specific residue and restore the original snapshot afterwards.


```python
# Focus on residue index 20
view.camera.zoom(selection='group_index==20')
```

```python
# Return to the original viewpoint snapshot
view.camera.set_snapshot(my_camera)
```
