# Unit 09: Granular Representations

Controlling the visual representations and transparencies.

### Learning Objectives
1. Apply cartoon, licorice, and spacefill representations.
2. Fine-tune representation properties like radius and detail.
3. Control region transparencies with the alpha parameter.

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Changing Representations on Whole

Let's change the global representation style.


```python
# Set global representation to spacefill with high transparency
view.whole.set_representation(representation='spacefill', alpha=0.2)
```

## 3. Styling a specific Region

We can render another part of the system as solid rods (licorice).


```python
# Create a region for Tyrosines
tyrosines = view.new_region(selection='res_name=="TYR"', tag='tyrosines')
tyrosines.set_representation(representation='licorice', alpha=1.0)
```
