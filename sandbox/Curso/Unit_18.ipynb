# Unit 18: Time & Trajectories

Navigating time-resolved structure trajectories.

### Learning Objectives
1. Load molecular dynamics trajectories.
2. Control play, pause, and speed settings.
3. Go to specific frame structure indexes.

## 1. Setting the Stage: Loading Trajectories

We load the pre-packaged trajectory demo for Pentalanine.


```python
import molsysviewer as msv

# Load trajectory demo
view = msv.demo['pentalanine']
view
```

## 2. Playback Controls

We can control the animation speed and loop states.


```python
# Play the dynamics at 20 frames per second
view.player.play(fps=20)
```

```python
# Pause the playback
view.player.pause()
```

## 3. Jumping to Frames

We can navigate to a specific frame directly.


```python
# Go to frame 30 (index 29)
view.player.go_to_structure(index=29)
```
