# Unit 14: Add-ons & Extensibility

MolSysViewer is built to be a platform, not just a standalone application. Through the **Add-on System**, you can expand its capabilities with specialized tools for Elastic Networks, Pharmacophores, Topology Analysis, and more. In this unit, we will learn how to discover and activate these "superpowers".

### Learning Objectives
1. Discover available add-ons in your environment.
2. Enable and disable add-ons for a specific viewer session.
3. Understand how add-ons transform the User Interface (Workspaces and Panels).

## 1. Discovering Add-ons

Add-ons are typically separate Python packages (e.g., `molsysviewer_elasnetmt`). The global `msv.addons` registry manages their detection.

```python
import molsysviewer as msv

# See which add-ons are installed and registered
print(f"Available add-ons: {msv.addons.available()}")

# If you just installed one, you can run a discovery scan
# msv.addons.discover()
```

## 2. Activating an Add-on

Once an add-on is discovered, you can enable it for your current viewer. This triggers the add-on's internal initialization, which may add new panels or visual objects.

```python
view = msv.MolSysView()
view.load('1CRN')

# For this lesson, we will use a template add-on as an example
from molsysviewer.addon_templates import minimal_elasnetmt
msv.addons.register_module(minimal_elasnetmt)

# Enable the add-on for this view
view.addons.enable('elasnetmt-template')

view
```

## 3. UI Transformation

Notice how the **Workbench** panel might have changed or disappeared. Add-ons often define their own **Workspace**, providing a tailored set of tools and removing generic ones to focus your workflow.

```python
# Check which add-ons are active in this view
print(f"Enabled in this view: {view.addons.enabled()}")

# You can also inspect the detailed contributions of an add-on
view.addons.records()[0] # metadata about panels, shapes, and tools
```

### Summary
The Add-on System allows MolSysViewer to evolve into a specialized workstation.
- Use `msv.addons.available()` to see what's in the box.
- Use `view.addons.enable()` to activate specialized UI and logic.
- Be prepared for UI changes when switching into an add-on Workspace.
