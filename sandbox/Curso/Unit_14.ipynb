# Unit 14: Textual Annotations (Labels)

Adding label annotations anchored to selections.

### Learning Objectives
1. Add textual labels to selections.
2. Label active selections directly.
3. Show, hide, and clear label annotations.

## 1. Setting the Stage

Load Crambin.


```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Adding Text Annotations

We use `add_annotation` to anchor labels to residues.


```python
# Add an annotation tag 'site_A' at residue 10
view.annotations.add_annotation(text='Catalytic Residue', selection='group_index==10', tag='site_A')

# List active tags
print("Active annotations:", view.annotations.tags)
```

## 3. Operations on Annotations

We can hide, display, or delete annotations programmatically.


```python
# Hide the label
view.annotations['site_A'].hide()
```

```python
# Remove all annotations
view.annotations.clear()
```
