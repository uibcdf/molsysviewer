# Unit 03: Structural Basics: Whole & Regions

Once we have loaded a structure and learned how to select its parts, the next step is understanding the **Visual Hierarchy**. In MolSysViewer, we manage the scene through two main concepts: the **Whole** system and individual **Regions**.

### Learning Objectives
1. Control the global structural baseline using the `whole` object.
2. Create custom, named **Regions** to isolate specific parts of the molecule.
3. Master the independent visibility of structural components.

## 1. Setting the Stage

Let's load a protein to practice our structural organization.

```python
import molsysviewer as msv
view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. The Global Baseline: `view.whole`

The `whole` object represents every atom currently loaded in the viewer. It is your "master control". Any command sent to `whole` affects the entire system unless overridden by a more specific region.

```python
# Hide everything to start from a blank canvas
view.whole.hide()
```

```python
# Show it again
view.whole.show()
```

## 3. Creating Regions: Anatomical Segments

Regions are named subsets of atoms. They allow us to assign a semantic identity to a part of the molecule (e.g., "Helix 1", "Active Site", "Chain A"). You can create them manually or use automatic tools.

```python
# Manual creation: Define the first 10 residues
core = view.new_region(selection='group_id<10', tag='nucleo')

print(f"Manual region: {list(view.regions.keys())}")
```

### Automatic Segmentation

For multi-chain systems, you don't need to define every chain one by one. The `make_regions_by()` method can automatically segment the structure based on its biological hierarchy.

```python
# Automatically create one region for every protein chain
view.make_regions_by(element='chain')

print(f"All regions (including automatic ones): {list(view.regions.keys())}")
```

## 4. Independent Visibility

Now that we have regions, we can manipulate them separately. This is the foundation of structural storytelling: showing only what matters for your analysis.

```python
view.whole.hide()
```

```python
view.regions['A'].hide()
```

```python
view.regions['nucleo'].hide()
```

```python
view.regions['nucleo'].show()
```

```python
# Focus the camera specifically on our new region
view.regions['nucleo'].focus()
```

### Summary
Structural basics are about organizing the "what" of your scene.
- `view.whole` controls the entire system.
- `view.new_region()` creates named, addressable subsets.
- Visibility can be toggled independently for every part.
