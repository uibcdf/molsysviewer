# Unit 13: Advanced Sectioning & Clipping

Dense molecular systems, such as large complexes or membrane proteins, often hide interesting features in their interior. In this unit, we will learn how to use **Clipping Planes** and **Advanced Sectioning** to "slice" through our models and reveal their internal dynamics.

### Learning Objectives
1. Control camera-space clipping with `set_clip_planes()`.
2. Perform world-space sectioning using `add_section()`.
3. Anchor cutting planes to structural features (Centroids).
4. Interact with the scene using 3D sectioning gizmos.

## 1. Setting the Stage

Let's load a protein and create a region for reference.

```python
import molsysviewer as msv
view = msv.MolSysView()
view.load('1CRN', label='crambin')
view

# Create a region for the core
#view.new_region(selection='10<=group_id<20', tag='active_site')
#view
```

## 2. Camera-Space Clipping

The simplest form of clipping is relative to the camera. Use `set_clip_planes()` to cut the front part of the scene based on a percentage (0-100) of the scene's radius.

```python
# Clip the first 50% of the front view
view.scene.set_clip_planes(near=50.0)
```

```python
# Reset clipping
view.scene.set_clip_planes(near=0.0)
```

```python
view.scene.set_clip_planes(near=50.0, far=True)
```

## 3. Advanced World-Space Sectioning

Unlike camera clipping, **Sections** are fixed to the molecular coordinates. You can place a cutting plane at an exact position and orientation. 

The `add_section()` method can anchor a plane to the **centroid** of any existing scene object (region, shape, or annotation).

```python
# Add a cutting plane exactly at the center of our 'active_site' region
# The normal [0, 0, 1] means we cut along the Z-axis
active_site = view.new_region(selection='10<=group_id<20', tag='active_site')
centroid = active_site.get_center()

my_cut = view.scene.add_section(
    point=centroid, 
    normal=[0, 0, 1], 
    tag='core_section'
)

print(f"Active sections: {view.scene._view._section_history}")
```

## 4. Interaction with Gizmos

When a section is added, you will see a semi-transparent disc in the 3D canvas. This is an **Interactive Gizmo**. 

**Instructions:**
1. Look at the disc on the canvas.
2. Drag the cyan handle in the center to move the plane along its axis.
3. Drag the yellow handle on the rim to rotate the cutting angle.

You can disable these visual handles without removing the cut using `disable_drag()`.

```python
# Hide the interactive handles for a clean view
view.scene._view._scene_objects['core_section'].disable_drag()

# To remove the section completely:
# view.scene.remove_section('core_section')
```
