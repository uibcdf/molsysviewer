# Unit 08: Precision Measurements

Visualization is not just about looking; it's about measuring. In this unit, we will learn how to extract quantitative data—such as distances, angles, and dihedrals—directly from our 3D model. We will also explore how **MolSysViewer** integrates with **PyUnitWizard** to provide unit-safe physical quantities.

### Learning Objectives
1. Create measurements interactively and programmatically.
2. Extract numerical results from the measurement registry.
3. Work with physical quantities and units (Å, degrees).

## 1. Setting the Stage

Let's load a system to perform our measurements.

```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Interactive Measurements (UX)

You can create measurements directly in the canvas using the context menu.

**Instructions:**
1. Right-click on an atom and select **"Distance"**.
2. Click on a second atom to complete the measurement.
3. Observe the dashed line and the value appearing in the viewer.

## 3. Programmatic Measurements (API)

For reproducible analysis, we use the `measurements` manager. We can define endpoints using MolSysMT selection strings.

```python
# Measure the distance between two specific atoms
dist = view.measurements.add_distance(selection_a='atom_index==10', selection_b='atom_index==50', tag='my_dist')
```

```python
# Measure an angle between three atoms
angle = view.measurements.add_angle(selection_a='atom_index==10', selection_b='atom_index==11', selection_c='atom_index==12',
                                    tag='my_angle')
```

## 4. Extracting Quantitative Data

The results of our measurements are stored in the registry. We can retrieve them as Python objects, and they automatically include their physical units.

```python
# Get detailed information about all active measurements
info = view.measurements.info()
```

```python
info
```

```python
# Extract the value of our distance measurement
if len(info) > 0:
    d_val = info[0].get('value')
    print(f"Distance Result: {d_val}")
    # Note: d_val is a PyUnitWizard quantity (e.g., 12.5 Å)
```

## 5. Navigation and Cleaning

Individual measurement objects support the `focus()` method to quickly move the camera to the measured feature.

```python
# Focus the camera on the distance we just measured
view.measurements['my_dist'].focus()
```

```python
view.measurements['my_dist'].delete()
```

```python
# Remove all measurements from the scene
view.measurements.clear()
```

### Summary
Measurements transform the viewer into an analytical workstation.
- Use `add_distance`, `add_angle`, or `add_dihedral` for programmatic control.
- Retrieve values using `info()`. 
- Always leverage **PyUnitWizard** for scientifically accurate results.
