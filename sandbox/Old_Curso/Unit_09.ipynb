# Unit 09: Geometric Shapes and Layers

Molecular analysis often requires visualizing concepts that are not physical atoms, such as centers of mass, binding volumes, or spatial directions. In this unit, we will learn how to add **Geometric Shapes** to our scene, define their dimensions using precise physical units, and organize them into logical **Layers**.

### Learning Objectives
1. Add abstract objects (spheres) to the 3D scene.
2. Master the **Unified Shapes API** (adding one or many shapes).
3. Understand the hierarchy between **Shapes** (individual identity) and **Layers** (visibility groups).
4. Use **PyUnitWizard** for unit-safe spatial coordinates.

## 1. Setting the Stage

Let's load a protein and introduce **PyUnitWizard**, which MolSysViewer uses to handle physical quantities safely.

```python
import molsysviewer as msv
import pyunitwizard as puw
import numpy as np

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Adding and Inspecting Shapes

We can place spheres at specific coordinates. Notice that we provide coordinates and radii as **Quantities** (strings or unit objects).

```python
# Add a large, semi-transparent sphere at the origin
marker = view.shapes.add_sphere(
    center='[0.0, 0.0, 0.0] nm', 
    radius='1.0 nm', 
    color=0xFF0000, 
    alpha=0.4, 
    tag='origin_marker'
)

# You can retrieve geometric data back from the object
print(f"Marker center: {marker.get_center()}")
```

## 3. Modifying Properties

Shapes are not static. You can update their appearance or navigate to them programmatically.

```python
# Change the marker's color to green
view.shapes['origin_marker'].set_color(0x00FF00)

# Focus the camera on this shape
view.shapes['origin_marker'].focus()
```

## 4. Batch Operations and Layering

When adding multiple shapes in a single call, MolSysViewer automatically organizes them. You can provide a `layer_tag` to group them logically for visibility management.

```python
# Define multiple centers
centers = puw.quantity([[2.0, 0.0, 0.0], [0.0, 2.0, 0.0], [0.0, 0.0, 2.0]], 'nm')

# Add them as a batch within a specific layer group
view.shapes.add_sphere(
    center=centers, 
    radius='0.3 nm', 
    color=0x0000FF, 
    tag='axis', 
    layer_tag='coordinate_system'
)
```

## 5. Registry and Visibility Control

The `view.shapes` registry allows you to list all geometric objects. To manage groups of objects, we use the `view.layers` registry.

```python
# List all shapes currently in the scene
print(f"Shapes in scene: {list(view.shapes.keys())}")
```

```python
# Hide the entire coordinate system layer without deleting the shapes
view.layers['coordinate_system'].hide()
```
