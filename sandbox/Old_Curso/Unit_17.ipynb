# Unit 17: Structural Comparison (Superposition)

Identifying the differences between two protein conformations or homologs requires aligning them in 3D space. In this unit, we will learn how to perform **Structural Superposition** using the synergy between **MolSysMT** (for calculation) and **MolSysViewer** (for visualization).

### Learning Objectives
1. Load multiple structures into a single viewer session.
2. Use MolSysMT tools to align structures programmatically.
3. Highlight conformational changes using different styles and colors.

## 1. Setting the Stage

Let's load two different structures of the same protein (e.g., from an MD trajectory or different PDB entries) to compare them.

```python
import molsysviewer as msv
import molsysmt as msm

view = msv.MolSysView()
# Load initial structure
view.load('1CRN', label='state_A')
# Load a second structure (or another frame) in 'add' mode
view.load('1CRN', label='state_B', mode='add')

# Shift state_B slightly so we can see they are not aligned yet
msm.basic.translate(view._molsys, translation=[1.0, 0, 0], selection='region:state_B')

view
```

## 2. Programmatic Superposition

We use MolSysMT's `align()` or `superpose()` functions to calculate the optimal transformation matrix and apply it to the coordinates of our system. The viewer will update automatically.

```python
# Align state_B onto state_A using Alpha Carbons as reference
msm.structure.align(view._molsys, selection='region:state_B', reference_selection='region:state_A', reference_structure_index=0)

# The viewer coordinates are now updated. Both structures occupy the same space.
```

## 3. Visualizing Differences

To make the comparison meaningful, we should style the two states differently. This allows us to see exactly where the structure fluctuates.

```python
# Color state_A in grey and state_B in orange
view.regions['state_A'].set_color_by_values([0xAAAAAA], element='system')
view.regions['state_B'].set_color_by_values([0xFFA500], element='system')

# Set state_A to a simple trace and state_B to a full cartoon
view.regions['state_A'].set_representation(representation='backbone')
view.regions['state_B'].set_representation(representation='cartoon')
```

### Summary
Structural comparison is a hybrid workflow in the UIBCDF ecosystem.
- Use **MolSysMT** for the mathematical alignment of coordinates.
- Use **MolSysViewer** to inspect the overlay and conformational changes.
- Differentiate states using regions, colors, and styles.
