# Unit 14: Add-ons & Extensibility

> **Important:** This unit requires the `elasnetmt` package. You can install it via: `pip install molsysviewer_elasnetmt` (or from the local repository).

MolSysViewer is an extensible platform. In this unit, we will explore the **ElasNetMT** add-on, learning how to operate it through the UI and, more importantly, how to use its **Public Python API** for direct structural analysis.

### Learning Objectives
1. Enable and discover scientific add-ons.
2. Navigate specialized **Add-on Workspaces**.
3. Use the **Add-on's own Python library** to drive the visualization programmatically.

## 1. Initializing with Add-ons

Add-ons are separate packages that register themselves with the viewer. Let's load a protein and check if `elasnetmt` is available.

```python
import molsysviewer as msv
view = msv.MolSysView()
view.load('1CRN', label='protein')
```

```python
msv.addons.available()
```

```python
# Enable the elastic network add-on
view.addons.enable('elasnetmt')
```

```python
view
```

## 2. The Power of Direct API Manipulation

While you can use the right-click menu or the side panels, add-ons often expose a public Python API. This is the most powerful way to use them, as it allows for custom parameters and automated workflows.

```python
# Import tools directly from the add-on package
from molsysviewer_elasnetmt.adapters import render_contact_network, render_anisotropy_ellipsoids

# Render the elastic network with custom parameters directly in our current view
render_contact_network(
    view, 
    cutoff='10.0 angstroms', 
    color='orange', 
    alpha=0.5,
    tag='my_enm'
)

print("Elastic network rendered using the Add-on's native API!")
```

## 3. High-level Context Actions

The viewer also provides a generic bridge to trigger add-on features without importing their libraries. This is useful for building generic UI wrappers.

```python
# Trigger a different analysis via the Viewer's Add-on bridge
view.addons.handle_context_action(
    addon='elasnetmt', 
    action_id='show-anisotropy-ellipsoids', 
    payload={}
)
```

### Summary
Add-ons are not just buttons; they are integrated Python ecosystems.
- Use `view.addons.enable()` to unlock the UI components.
- Use the **Add-on's internal library** (e.g., `molsysviewer_elasnetmt.adapters`) for maximum programmatic control.
- Combine both worlds to build tailored biophysical analysis dashboards.
