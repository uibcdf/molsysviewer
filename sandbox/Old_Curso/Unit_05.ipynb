# Unit 05: Presets and Smart Views (The Recipes)

In the previous unit, we learned how to build a scene using manual "bricks". However, achieving a professional scientific look often requires combining several styles simultaneously. In this unit, we will explore **Presets**: high-level Mol* recipes that automate the visual composition for us.

### Learning Objectives
1. Understand the conceptual difference between a **Representation** and a **Preset**.
2. Use standard smart views like `atomic-detail` and `polymer-cartoon`.
3. Learn how to load custom **User Presets** from external configuration files.

## 1. Setting the Stage

For this unit, let's load a more complex system: a protein dimer (`1TCD`).

```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1TCD', label='dimer')
view
```

## 2. Using Built-in Recipes

Instead of manually adding sticks and spheres, we can apply the `atomic-detail` preset. This recipe automatically identifies bonds and atoms to create a classic chemical model.

```python
# Apply the 'atomic-detail' preset to the whole system
view.whole.set_representation(preset='atomic-detail')
```

Another common recipe is `polymer-cartoon`, which is optimized for providing a clear view of protein secondary structure and nucleic acid topology.

```python
# Revert to a clean, professional cartoon view
view.whole.set_representation(preset='polymer-cartoon')
```

## 3. Custom User Presets

Power users can define their own visual recipes in YAML or JSON files. These files contain rules for coloring, sizing, and layering multiple representations. You can load these externally defined looks using the `load_user_presets` helper.
