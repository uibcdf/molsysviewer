# Unit 05: The Camera Manager

In previous units, we learned how to style our molecules. Now, we will learn how to navigate the 3D space around them. **MolSysViewer** features a dedicated `CameraManager` that allows for precise, programmatic control over your viewpoint.

### Learning Objectives
1. Center and zoom the view on specific structural features.
2. Switch between **Perspective** and **Orthographic** projection modes.
3. Capture and restore **Camera Snapshots** to maintain consistent viewpoints.

## 1. Setting the Stage

First, let's load a protein structure to explore.

```python
import molsysviewer as msv

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Dynamic Zooming

The `zoom()` method is your primary tool for navigation. You can pass a MolSysMT selection string to focus the camera on a specific part of the system. The `duration` parameter allows for smooth, cinematic transitions.

## 3. Projection Modes

By default, the viewer uses **Perspective** projection (things get smaller as they move away). For scientific figures, you might prefer **Orthographic** projection, which removes distortion and is ideal for architectural or front-on views.

```python
# Switch to orthographic mode for a flatter, technical look
view.camera.set_mode('orthographic')
```

```python
# Switch back to perspective for immersion
view.camera.set_mode('perspective')
```

## 4. Camera Snapshots

Once you find the "perfect angle," you can save it as a snapshot. This is incredibly useful for reproducibility: you can move around freely and jump back to your saved viewpoint with a single command.

```python
# Capture the current camera state
my_viewpoint = view.camera.get_snapshot()
```

```python
my_viewpoint
```

Snapshot captured. Move the camera manually, then run the next cell to return.

```python
# Restore the saved viewpoint
view.camera.set_snapshot(my_viewpoint)
```

### Summary
The `CameraManager` separates navigation from content. 
- Use `zoom()` to focus on details.
- Use `mode` to define the projection style.
- Use `snapshots` to bookmark your work.
