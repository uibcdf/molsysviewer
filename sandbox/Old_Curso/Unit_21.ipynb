# Unit 21: Coordinate Engineering (Dynamic Editing)

Molecular structures in **MolSysViewer** are not just for display; they are dynamic entities that can be manipulated programmatically. In this unit, we will learn how to extract, perturb, and re-inject atomic coordinates into the viewer to simulate structural changes in real-time.

### Learning Objectives
1. Extract current atomic positions using `get_coordinates()`.
2. Apply spatial transformations (translations, perturbations) in Python.
3. Update the 3D scene instantly using `set_coordinates()`.

## 1. Setting the Stage

Let's load a simple protein to perform our engineering experiments.

```python
import molsysviewer as msv
import pyunitwizard as puw
import numpy as np

view = msv.MolSysView()
view.load('1CRN', label='crambin')
view
```

## 2. Extracting Coordinates

We can retrieve the positions of all atoms (or a selection) as a **PyUnitWizard** quantity. This data is a live snapshot of the current structure.

```python
# Get coordinates of all atoms in nanometers
coords = view.get_coordinates(selection='all')

print(f"Coordinate array shape: {coords.shape}")
print(f"First atom position: {coords[0, 0]}")
```

## 3. Applying Perturbations

Using standard Python tools (NumPy), we can modify these coordinates. For example, let's shift a specific residue by 0.5 nm in the X direction.

```python
# Extract coordinates of residue 10
indices = view.select('res_id 10')
res_coords = view.get_coordinates(selection=indices)

# Apply a translation in Python
new_res_coords = res_coords + puw.quantity([0.5, 0, 0], 'nm')

# Push the updated coordinates back to the viewer
view.set_coordinates(new_res_coords, selection=indices)

print("Structure updated! Observe the displacement in the viewer.")
```

## 4. Engineering Shapes

The same principle applies to non-structural objects. You can move geometric shapes by redefining their coordinate payload.

```python
# Add a marker sphere
marker = view.shapes.add_sphere(center=[0,0,0], radius='0.5 nm', tag='target')

# Move the marker to a new position programmatically
marker.set_coordinates(puw.quantity([1.0, 1.0, 1.0], 'nm'))
```

### Summary
Coordinate engineering enables interactive modeling and conformational analysis.
- Use `get_coordinates()` to pull data into Python.
- Manipulate positions using NumPy and physical units.
- Use `set_coordinates()` to push changes to the GPU render tree.
