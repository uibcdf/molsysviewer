Módulo 11: Compartiendo Descubrimientos (Exportación de Imágenes y HTML)

  Objetivos Pedagógicos:
   1. Generar imágenes PNG de alta resolución.
   2. Exportar la sesión completa a un archivo HTML portable que funcione sin Python.
   3. Entender la diferencia entre exportación "Standalone" y "Lite".

```python
import molsysviewer as msv
view = msv.MolSysView()
view.load('1CRN')
```

```python
view
```

```python
view.export.image('mi_proteina.png', width_px=1600, height_px=1200, scale=2.0, transparent=True)
```

```python
view.export.html('mi_sesion.html', title="Análisis de Crambin", mode='standalone')
```
