# Unit 24: View Cloning & Surgical Extraction

In the final unit of this course, we will learn how to manage the lifecycle of our visualization sessions. We will explore how to duplicate entire views and, more importantly, how to perform **Surgical Extractions**: creating focused sub-views that preserve all your styles and annotations while filtering out structural noise.

### Learning Objectives
1. Duplicate a session exactly using `copy()`.
2. Extract a sub-view from a selection using the intelligent `extract()` tool.
3. Understand how visual objects (Regions, Shapes, Measurements) survive and remap during extraction.

## 1. Setting the Stage

Let's create a complex scene with multiple annotations and regions to test our extraction capabilities.

```python
import molsysviewer as msv
view = msv.MolSysView()
view.load('1CRN', label='crambin')

# Add some visual context
view.new_region(selection='res_id 10:20', tag='core_region', representation='licorice')
view.annotations.add_label(text='Important Atom', selection='atom_index 100', tag='label1')
view.measurements.add_distance(atom_indices_a=[50], atom_indices_b=[60], tag='dist1')
view
```

## 2. Intelligent Extraction

The `extract()` tool is much more than a structural filter. It identifies which of your visual objects are relevant to the extracted subset and migrates them automatically, remapping atom indices in the process.

```python
# Extract only the 'core' residues (10 to 20)
# This will return a new, independent viewer
sub_view = view.extract(selection='res_id 10:20')

# Display the extracted view
sub_view
```

## 3. Auditing Survival

Let's check what made it into the new view. 
- The `core_region` should survive (because its atoms are in the extract).
- The `label1` might be dropped if atom 100 was not in the selection.
- The `dist1` will be dropped if atoms 50 or 60 are absent.

```python
print(f"Original regions: {list(view.regions.keys())}")
print(f"Extracted regions: {list(sub_view.regions.keys())}")

print(f"Extracted annotations: {sub_view.annotations.tags()}")
```

## 4. Full Session Cloning

If you just want an identical copy of your work to test a different styling path without affecting the original, use `copy()`.

```python
# Create an exact twin of the current view
view_copy = view.copy()

# Any change in view_copy will NOT affect the original 'view'
view_copy.whole.set_representation(representation='spacefill')
```
